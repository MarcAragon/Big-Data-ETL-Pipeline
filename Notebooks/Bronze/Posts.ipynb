```python
#2023
import dlt
import pyarrow.parquet as pq
import fsspec
import toml #Esto para forzar que se lea de aqui

secrets_path = "./.dlt/secrets.toml"
secrets = toml.load(secrets_path)
# Define un recurso que lee datos de un parquet remoto
@dlt.resource(table_name="Posts_2023")
def my_df():
    parquet_url = "https://datasets-documentation.s3.eu-west-3.amazonaws.com/stackoverflow/parquet/posts/2023.parquet"
    with fsspec.open(parquet_url, mode="rb") as f:
        parquet_file = pq.ParquetFile(f)
        print('Extrayendo parquet')

        print(f"Total de filas: {parquet_file.metadata.num_rows}")
        print(f"Número de row groups: {parquet_file.num_row_groups}")

        for i, batch in enumerate(parquet_file.iter_batches()):
            print(f"Procesando batch {i + 1}...")

            df = batch.to_pandas()
            df = df.convert_dtypes()

            print(f"Batch {i + 1} con {len(df)} filas")

            yield df


# 🚀 Pipeline configurado con destino filesystem
# Las credenciales se leerán automáticamente desde `.dlt/secrets.toml`
pipeline = dlt.pipeline(
    pipeline_name="Bronze_MinIO",
    destination="filesystem",
    dataset_name="Posts/2023"
)

# Ejecutamos el pipeline
load_info = pipeline.run(
    my_df,
    loader_file_format="parquet",
    write_disposition="replace"
)
print(load_info)
```
